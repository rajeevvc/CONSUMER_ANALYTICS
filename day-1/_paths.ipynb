# Shared — `_paths.py`

Tiny helper that exposes `DAY1`, `DATA`, and `OUTPUT` path constants.


```python
from pathlib import Path
import sys

HERE = Path.cwd()
DAY1 = HERE / "day-1" if (HERE / "day-1").exists() else HERE
if str(DAY1) not in sys.path:
    sys.path.insert(0, str(DAY1))

import _paths
print("DAY1  =", _paths.DAY1)
print("DATA  =", _paths.DATA)
print("OUTPUT=", _paths.OUTPUT)
```
