## Setup


```python
from __future__ import annotations

from pathlib import Path
import sys

HERE = Path.cwd()
if (HERE / "day-1").exists():
    _DAY1_ROOT = HERE / "day-1"
elif HERE.name in {"01_sessionize", "02_journey", "03_state_modeling", "04_feature_engineering", "05_segmentation", "_shared"}:
    _DAY1_ROOT = HERE.parent
elif HERE.name == "examples":
    _DAY1_ROOT = HERE.parents[1]
else:
    _DAY1_ROOT = HERE

# Make Path(__file__) match the companion script (same as running the .py).
__file__ = str(_DAY1_ROOT / "_paths.py")
print("DAY1 root =", _DAY1_ROOT)
print("__file__ =", __file__)
```

## Imports & constants


```python
from pathlib import Path

DAY1 = Path(__file__).resolve().parent
DATA = DAY1 / "data"
OUTPUT = DAY1 / "output"
SHARED = DAY1 / "_shared"

CONTEXT_DIRS = {
    "sessionize": DAY1 / "01_sessionize",
    "journey": DAY1 / "02_journey",
    "state_modeling": DAY1 / "03_state_modeling",
    "feature_engineering": DAY1 / "04_feature_engineering",
    "segmentation": DAY1 / "05_segmentation",
}
```

## Helpers (one function per cell)


### `ensure_day1_on_path`


```python
def ensure_day1_on_path() -> Path:
    """Insert day-1 onto sys.path so sibling context packages can import."""
    import sys

    day1 = str(DAY1)
    if day1 not in sys.path:
        sys.path.insert(0, day1)
    return DAY1
```

### `add_context_to_path`


```python
def add_context_to_path(context_key: str) -> Path:
    """Allow `from step0N_... import ...` across context folders."""
    import sys

    ensure_day1_on_path()
    folder = CONTEXT_DIRS[context_key]
    path = str(folder)
    if path not in sys.path:
        sys.path.insert(0, path)
    return folder
```

## Run


```python
print('DAY1 =', DAY1)
print('on path:', ensure_day1_on_path())
```
